2.  **Model Building**: Utilizes TensorFlow/Keras with EfficientNetB0 for transfer learning.
3.  **Training**: Configures and trains the model with callbacks like `EarlyStopping`, `ReduceLROnPlateau`, and `ModelCheckpoint`.

5.  **Artifact Saving**: Saves the trained model, training history, and plots.
6.  **Prediction Function**: Provides a utility to predict class and confidence for new images.
7.  **Fine-tuning**: Optimizes the model for higher validation accuracy.
8.  **Production Readiness**: Designed for clarity and suitability for a GitHub portfolio.

### Dataset Detection and Initial Preprocessing

First, we need to locate the dataset. Colab typically uploads files to the `/content/` directory. We'll search for common dataset structures or zip files and extract them if found. Then, we'll load the data and display essential statistics before proceeding with training.

````python
import os

# Directly set the DATASET_PATH to the expected location
DATASET_PATH = '/content/blood_cancer_dataset'

print(f"Using dataset path: {DATASET_PATH}")
````

````python
# --- Data Augmentation Layer ---
data_augmentation = tf.keras.Sequential([
    tf.keras.layers.RandomFlip("horizontal_and_vertical"),
    tf.keras.layers.RandomRotation(0.2),
    tf.keras.layers.RandomZoom(0.2),
    tf.keras.layers.RandomContrast(0.2),
    tf.keras.layers.RandomBrightness(0.2)
])

print("Data augmentation layer defined.")
````

````python
IMG_HEIGHT = 224
IMG_WIDTH = 224
BATCH_SIZE = 32

# --- Count images per class and overall ---
# This step is done manually to fulfill the 'number of images per class' requirement
# before splitting, and to ensure we know the true class names.

class_names = sorted([d for d in os.listdir(DATASET_PATH) if os.path.isdir(os.path.join(DATASET_PATH, d))])
if not class_names:
    # If the root is directly containing train/val/test, then look inside 'train' or 'data'
    for sub_dir in ['train', 'data']:
        if os.path.isdir(os.path.join(DATASET_PATH, sub_dir)):
            class_names = sorted([d for d in os.listdir(os.path.join(DATASET_PATH, sub_dir)) if os.path.isdir(os.path.join(DATASET_PATH, sub_dir, d))])
            if class_names:
                DATASET_PATH = os.path.join(DATASET_PATH, sub_dir) # Adjust base path to the one containing classes
                print(f"Adjusted dataset root to: {DATASET_PATH}")
                break

if not class_names:
    raise ValueError("No class subdirectories found in the detected dataset path.")

image_counts_per_class = {}
total_images = 0
for class_name in class_names:
    class_dir = os.path.join(DATASET_PATH, class_name)
    if os.path.isdir(class_dir):
        num_images = len([f for f in os.listdir(class_dir) if f.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp', '.gif'))])
        image_counts_per_class[class_name] = num_images
        total_images += num_images

# Print dataset statistics
print(f"\nDataset path: {DATASET_PATH}")
print(f"Number of classes: {len(class_names)}")
print(f"Class names: {class_names}")
print("Number of images per class:")
for cls, count in image_counts_per_class.items():
    print(f"  - {cls}: {count} images")
print(f"Total images in dataset: {total_images}")

# --- Create datasets with 80/10/10 split ---
# Using 0.2 for validation_split will create 80% train, 20% val/test combined.
# We will split this 20% further into 10% validation and 10% test.

# Step 1: Create training and combined validation/test datasets
train_ds = tf.keras.utils.image_dataset_from_directory(
    DATASET_PATH,
    validation_split=0.2,
    subset="training",
    seed=SEED,
    image_size=(IMG_HEIGHT, IMG_WIDTH),
    batch_size=BATCH_SIZE,
    label_mode='categorical' # Use categorical for EfficientNetB0 output
)

val_test_ds = tf.keras.utils.image_dataset_from_directory(
    DATASET_PATH,
    validation_split=0.2,
    subset="validation",
    seed=SEED,
    image_size=(IMG_HEIGHT, IMG_WIDTH),
    batch_size=BATCH_SIZE,
    label_mode='categorical'
)

# Step 2: Split val_test_ds into true validation and test sets (50/50 of the 20% = 10% each)
val_batches = tf.data.experimental.cardinality(val_test_ds).numpy() // 2
validation_ds = val_test_ds.take(val_batches)
test_ds = val_test_ds.skip(val_batches)

# Calculate and print sample counts for splits
num_train_samples = tf.data.experimental.cardinality(train_ds).numpy() * BATCH_SIZE
num_val_samples = tf.data.experimental.cardinality(validation_ds).numpy() * BATCH_SIZE
num_test_samples = tf.data.experimental.cardinality(test_ds).numpy() * BATCH_SIZE

print(f"\nTraining samples: {num_train_samples}")
print(f"Validation samples: {num_val_samples}")
print(f"Test samples: {num_test_samples}")

# Configure the datasets for performance
def configure_for_performance(ds):
    ds = ds.cache()
    ds = ds.prefetch(buffer_size=tf.data.AUTOTUNE)
    return ds

train_ds = configure_for_performance(train_ds)
validation_ds = configure_for_performance(validation_ds)
test_ds = configure_for_performance(test_ds)

print("\nDataset loading and initial configuration complete.")
````

### Build and Compile EfficientNetB0 Model

````python
# Load the EfficientNetB0 model pre-trained on ImageNet
baseline_model = tf.keras.applications.EfficientNetB0(include_top=False, weights='imagenet', input_shape=(IMG_HEIGHT, IMG_WIDTH, 3))

# Freeze the pre-trained layers
baseline_model.trainable = False

# Create a new model on top of the pre-trained base
inputs = keras.Input(shape=(IMG_HEIGHT, IMG_WIDTH, 3))

# Apply data augmentation only to the training data (optional, but good practice)
x = data_augmentation(inputs)

# Preprocess inputs for EfficientNetB0
x = tf.keras.applications.efficientnet.preprocess_input(x)

x = baseline_model(x, training=False)
x = keras.layers.GlobalAveragePooling2D()(x)
x = keras.layers.Dropout(0.2)(x) # Add a dropout layer for regularization
outputs = keras.layers.Dense(len(class_names), activation='softmax')(x) # Output layer with softmax for multi-class classification

model = keras.Model(inputs, outputs)

# Compile the model
model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.001),
    loss='categorical_crossentropy',
    metrics=['accuracy', tf.keras.metrics.Precision(), tf.keras.metrics.Recall()]
)

model.summary()
````

### Configure Training Callbacks

````python
model_checkpoint_callback = tf.keras.callbacks.ModelCheckpoint(
    filepath='best_model.keras', # .keras format is recommended for Keras 3
    save_weights_only=False, # Save the entire model
    monitor='val_accuracy',
    mode='max',
    save_best_only=True,
    verbose=1
)

early_stopping_callback = tf.keras.callbacks.EarlyStopping(
    monitor='val_loss',
    patience=10, # Number of epochs with no improvement after which training will be stopped
    restore_best_weights=True, # Restores model weights from the epoch with the best value of the monitored quantity.
    verbose=1
)

reduce_lr_callback = tf.keras.callbacks.ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.2, # Factor by which the learning rate will be reduced. new_lr = lr * factor
    patience=5, # Number of epochs with no improvement after which learning rate will be reduced.
    min_lr=0.00001, # Lower bound on the learning rate.
    verbose=1
)

callbacks = [model_checkpoint_callback, early_stopping_callback, reduce_lr_callback]
````

### Train the Model

````python
EPOCHS = 30 # You can adjust this based on your dataset and early stopping

history = model.fit(
    train_ds,
    epochs=EPOCHS,
    validation_data=validation_ds,
    callbacks=callbacks
)
````

### 1. Evaluate Model on the Test Dataset

````python
print("Evaluating model on test dataset...")
loss, accuracy, precision, recall = model.evaluate(test_ds)

print(f"\nTest Loss: {loss:.4f}")
print(f"Test Accuracy: {accuracy:.4f}")
print(f"Test Precision: {precision:.4f}")
print(f"Test Recall: {recall:.4f}")

# Calculate F1 Score
if (precision + recall) > 0:
    f1_score = 2 * (precision * recall) / (precision + recall)
    print(f"Test F1 Score: {f1_score:.4f}")
else:
    print("Test F1 Score: 0.0000 (Precision and Recall are zero)")
````

### 5. Create Code to Upload and Predict a New Image

````python
from tensorflow.keras.preprocessing import image

def predict_single_image(image_path, model, class_names, img_height, img_width):
    img = image.load_img(image_path, target_size=(img_height, img_width))
    img_array = image.img_to_array(img)
    img_array = np.expand_dims(img_array, axis=0) # Create a batch

    # Preprocess the image for EfficientNetB0
    img_array = tf.keras.applications.efficientnet.preprocess_input(img_array)

    predictions = model.predict(img_array)
    predicted_class_index = np.argmax(predictions[0])
    predicted_class = class_names[predicted_class_index]
    confidence = np.max(predictions[0]) * 100

    return predicted_class, confidence

print("Prediction function `predict_single_image` created.")
print("You can now upload an image and call this function. For example:")
print("```python")
print("  # from google.colab import files")
print("  # uploaded = files.upload()")
print("  # for fn in uploaded.keys():")
print("  #    predicted_class, confidence = predict_single_image(fn, model, class_names, IMG_HEIGHT, IMG_WIDTH)")
print("  #    print(f\"Image: {fn}, Predicted: {predicted_class}, Confidence: {confidence:.2f}%\")")
print("```")
````
